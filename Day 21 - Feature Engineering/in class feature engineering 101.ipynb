***FEATURE ENGINEERING***

- it is the process of tranforming raw data into features that make machine learning algorithms work better.
    - to improve performance of ML models
    - it can improve quality and relevance of features

- **Catagorical Features**

    - Groups of data
        - will include Binary Features (0 and 1)
        - sex - M F
        - Survived - Yes NO
        - Bool data type
        - Category data type -
        - Ordinal data type - you can set up a scale in the categories - Small Medium Large - COld warm hot
        - small height medium height large height

- **Numerical Features**

    - where there are numbers
    - will include float and integers both
    - continous values,
    - grouping up will not make much sense (unless doing clustering)

```python
penguins = sns.load_dataset('penguins')
penguins.info()
```

- Encoding

```python
# label encoding
# column that you can encode

penguins['species'].unique()

# {  }

species_codes = {
    'Adelie': 0,
    'Chinstrap': 1,
    'Gentoo': 2
}

penguins['species_encoded'] = penguins['species'].map(species_codes)

penguins.sample(10)
```

```python
# pandas map
s = pd.Series(['cat', 'cow', 'dog'])

s2 = s.map({'cat': 'kitten', 'cow': 'calf'})
print(s2)
```

```python
# factorize
# species still contains the species names

# factorize will take the column and assigns to the new groups
x, y = pd.factorize(penguins['species'])

penguins [ 'species_factorized' ] = x
penguins
# penguins['species_factorize'],unique_penguins = pd.factorize(penguins['species'])
```

```python
factors = pd.factorize(penguins['species'])
penguins [ 'species_factorized' ] = factors[0]
penguins [[ 'species','species_factorized' ]]
```

- Using the **catagories** method

```python
df = penguins.copy()

df['species'] = df['species'].astype('category')

df['species_cat_encoded'] = df['species'].cat.codes
df
```
